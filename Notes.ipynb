## Token Organization

Create a tree-like structure that orders tokens by first letter, last letter, length. 

```python
{
    std::vector<tree_node<char>> tokens;

    for (unsigned char ch1 = 'a'; ch1 <= 'z'; ((unsigned int)ch1)++) {
        tree_node<char> a_token;
        a_token.t = ch1;
        tokens.push_back(a_token);
        for (unsigned char ch2 = 'a'; ch2 <= 'z'; ((unsigned int)ch2)++) {
            tree_node<char> z_token;
            z_token.t = ch2;
            a_token.children.push_back(z_token);

        }
    }
}
```

### Basic type agnostic tree.

```python
template< typename T >
struct tree_node
   {
   T t;
   std::vector<tree_node> children;
   };
```
